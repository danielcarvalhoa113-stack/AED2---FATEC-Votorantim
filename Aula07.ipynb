Projeto 1 - gestão de tarefas

```python
# Implementação simples de uma Fila (Queue)
class Queue:
    def __init__(self):
        self.itens = []

    def enqueue(self, item):
        self.itens.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.itens.pop(0) # Remove do início da fila
        return None

    def is_empty(self):
        return len(self.itens) == 0

def gestao_tarefas(tarefas_iniciais, ciclos, novas_por_ciclo=None):
    novas_por_ciclo = novas_por_ciclo or {}
    fila = Queue() # Agora a classe Queue está definida!

    for tarefa in tarefas_iniciais:
        fila.enqueue(tarefa)

    for ciclo in range(1, ciclos + 1):
        print(f"--- Ciclo {ciclo} ---")

        if fila.is_empty():
            print("Não há tarefa pendente para processar.")
        else:
            atual = fila.dequeue()
            print(f"Tarefa concluída: {atual}")

        for nova in novas_por_ciclo.get(ciclo, []):
            fila.enqueue(nova)
            print(f"Tarefa adicionada: {nova}")

# Exemplo
gestao_tarefas(
    ["Relatório", "E-mail", "Backup"],
    5,
    {1: ["Atualizar dashboard"],
     3: ["Conferir estoque"]}
)
```

Projeto 2 - Fila de cinema

```python
# Implementação da Fila (Queue) necessária para o código
class Queue:
    def __init__(self):
        self.itens = []

    def enqueue(self, item):
        self.itens.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.itens.pop(0)
        return None

    def is_empty(self):
        return len(self.itens) == 0

def fila_cinema(clientes_iniciais, total_atendimentos,
                atendimentos_por_sessao, chegadas_por_sessao=None):
    chegadas_por_sessao = chegadas_por_sessao or {}
    fila = Queue()  # Agora a classe Queue está definida corretamente

    for cliente in clientes_iniciais:
        fila.enqueue(cliente)

    atendidos = 0
    sessao = 1

    while atendidos < total_atendimentos:
        print(f"--- Sessão {sessao} ---")

        for novo in chegadas_por_sessao.get(sessao, []):
            fila.enqueue(novo)
            print(f"Chegou à fila: {novo}")

        if fila.is_empty():
            print("Não há clientes aguardando.")
            break

        for _ in range(atendimentos_por_sessao):
            if fila.is_empty() or atendidos == total_atendimentos:
                break

            cliente = fila.dequeue()
            print(f"{cliente} comprou ingresso e entrou.")
            atendidos += 1

        sessao += 1

# Exemplo
fila_cinema(
    ["Cliente A", "Cliente B", "Cliente C"],
    total_atendimentos=5,
    atendimentos_por_sessao=2,
    chegadas_por_sessao={
        2: ["Cliente D", "Cliente E"],
        3: ["Cliente F"]
    }
)
```
