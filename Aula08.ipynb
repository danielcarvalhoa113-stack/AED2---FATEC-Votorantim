Projeto 1 - Gerenciador de Playlist de Música

```python
class Music:
    def __init__(self, title, artist):
        self.title = title
        self.artist = artist

    def __str__(self):
        return f"{self.title} - {self.artist}"


class DoublyNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class MusicPlaylist:
    def __init__(self):
        self.head = None
        self.tail = None
        self.current_song_node = None

    def add_song(self, music):
        new_node = DoublyNode(music)

        if not self.head:
            self.head = new_node
            self.tail = new_node
            self.current_song_node = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    def remove_song(self, title):
        current = self.head

        while current:
            if current.data.title == title:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next

                if current.next:
                    current.next.prev = current.prev
                else:
                    self.tail = current.prev

                if self.current_song_node == current:
                    self.current_song_node = current.next if current.next else self.head

                return True

            current = current.next

        return False

    def play_next(self):
        if not self.current_song_node:
            return None

        if self.current_song_node.next:
            self.current_song_node = self.current_song_node.next
        else:
            print("Fim da playlist. Voltando para o início.")
            self.current_song_node = self.head

        return self.current_song_node.data if self.current_song_node else None

    def play_previous(self):
        if not self.current_song_node:
            return None

        if self.current_song_node.prev:
            self.current_song_node = self.current_song_node.prev

        return self.current_song_node.data

    def mostrar_atual(self):
        if self.current_song_node:
            print(f"Tocando agora: {self.current_song_node.data}")
        else:
            print("Playlist vazia.")

# Exemplo de uso
playlist = MusicPlaylist()
playlist.add_song(Music("Bohemian Rhapsody", "Queen"))
playlist.add_song(Music("Stairway to Heaven", "Led Zeppelin"))
playlist.add_song(Music("Hotel California", "Eagles"))

playlist.mostrar_atual()  # Deve tocar Bohemian Rhapsody

print("\nAvançando...")
playlist.play_next()
playlist.mostrar_atual()  # Deve tocar Stairway to Heaven

print("\nAvançando...")
playlist.play_next()
playlist.mostrar_atual()  # Deve tocar Hotel California

print("\nVoltando...")
playlist.play_previous()
playlist.mostrar_atual()  # Deve voltar para Stairway to Heaven
```

Projeto 2 - Histórico de Navegação do Browser

```python
class HistoryNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class BrowserHistory:
    def __init__(self):
        self.head = None
        self.current = None

    def visit_page(self, url):
        new_node = HistoryNode(url)

        if not self.head:
            self.head = new_node
            self.current = new_node
            return

        # Se voltamos no histórico, descarta primeiro o trecho futuro.
        if self.current.next:
            temp = self.current.next
            self.current.next = None

            while temp:
                next_temp = temp.next
                temp.prev = None
                temp.next = None
                temp = next_temp

        self.current.next = new_node
        new_node.prev = self.current
        self.current = new_node

    def go_back(self):
        if self.current and self.current.prev:
            self.current = self.current.prev
            return self.current.data
        return None

    def go_forward(self):
        if self.current and self.current.next:
            self.current = self.current.next
            return self.current.data
        return None

    def display_history(self):
        history = []
        current = self.head

        while current:
            history.append(current.data)
            current = current.next

        print("Histórico de Navegação:")
        for url in history:
            # Nota: usamos 'is' para comparar se é exatamente o mesmo nó,
            # evitando problemas se houver URLs duplicadas no histórico.
            marker = " <- atual" if current == self.current else "" # Ajustado para robustez
            # Ou mantendo pela lógica do código original:
            marker = " <- atual" if url == self.current.data and history.count(url) == 1 else ""
            # Uma forma mais segura de marcar o atual comparando nós diretamente:
            # (Vamos simplificar imprimindo direto no loop pelos nós)

    def display_history(self):
        current = self.head
        print("Histórico de Navegação:")
        while current:
            marker = " <- atual" if current == self.current else ""
            print(f"- {current.data}{marker}")
            current = current.next


# Exemplo de uso
history = BrowserHistory()
history.visit_page("www.fatecvotorantim.edu.br")
history.visit_page("www.google.com")
history.visit_page("www.youtube.com")

print(f"Voltar: {history.go_back()}")
print(f"Avançar: {history.go_forward()}")
print(f"Voltar novamente: {history.go_back()}")

# Estamos no Google; uma nova visita descarta YouTube do futuro.
history.visit_page("www.wikipedia.org")
history.display_history()
print(f"Tentar avançar: {history.go_forward()}")
```
