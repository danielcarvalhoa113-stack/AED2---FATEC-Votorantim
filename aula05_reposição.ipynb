* O construtor recebe o nome do titular e inicia o saldo em 0. • Use __saldo como atributo privado.
* Implemente depositar(valor), permitindo apenas valores positivos.
* Implemente consultar_saldo(), retornando o saldo atual.
* Crie uma conta, faça dois depósitos e mostre o saldo.

```python
#Primeiro definimos a classe
class Conta:
  def __init__(self, saldo):
    self.__saldo = 0

  def depositar(self, valor):
    if valor <= 0:
      print(f"Valor digitado incompátivel!")
    else:
      self.__saldo += valor
  def consultar(self):
    print(f"{self.__saldo}")
```

Exercício 7 — Veículo com velocidade controlada  [Média]<br> Modele um veículo inspirado no exemplo Automovel da aula, mas com regras completas de aceleração e frenagem.<br>
**Requisitos:**
* A classe Veiculo deve ter placa, velocidade_max e velocidade_atual; torne os três atributos privados.
* velocidade_atual começa em 0.
* Crie getters para placa e velocidade_max.
* Crie um setter para velocidade_max que aceite apenas valores positivos e nunca deixe a nova máxima menor que a velocidade atual.
* acelerar(incremento=10) aumenta a velocidade sem ultrapassar a máxima.
* frear(decremento=10) reduz a velocidade sem ficar negativa.
* Implemente __str__ para mostrar placa, velocidade atual e máxima.
* Demonstre o funcionamento com várias acelerações e frenagens.

```python
#Método construtor com os atributos
class Carro:
  def __init__(self, placa, velocidade_max, velocidade_atual):
    self.__placa = placa
    self.__velocidade_max = 150
    self.__velocidade_atual = 0

  #Métodos de acesso e configuração
  def get_placa(self):
    return self.__placa

  def get_velocidade_max(self):
    return self.__velocidade_max

  def set_velocidade_max(self, velocidade_max):
    if velocidade_max <= 0 or velocidade_max <= self.__velocidade_atual:
      print(f"Erro")
    else:
      self.__velocidade_max = velocidade_max

  #Aceleração e Frenagem
  def acelerar(self):
    a = 10
    if self.__velocidade_atual == self.__velocidade_max:
      print(f"Já estamos na velocidade máxima")
    else:
      self.__velocidade_atual += a

  def freiar(self):
    b = 10
    if self.__velocidade_atual == 0:
      print(f"Chegamos ao destino!")
    else:
      self.__velocidade_atual -= b

  def descricao_carro(self):
    print(f"{self.__placa}, alcança {self.__velocidade_max} e agora está a {self.__velocidade_atual} km/h")
```

```python
#Criando um carro
prisma = Carro("PIVA023", 170, 0)
prisma.get_placa()
```

```python
#Acelerando um carro
prisma.acelerar()
prisma.descricao_carro()
```

```python
#Freiando um carro
prisma.freiar()
prisma.descricao_carro()
```

```python
#Modificando o carro
prisma.set_velocidade_max(150)
prisma.get_velocidade_max()
```

* Use atributos privados para id, nome e preco.
* Crie getters para id, nome e preco.
* Crie set_preco(novo_preco), aceitando apenas valores maiores que zero.

```python
#Definindo classe
class Produto:
  contador = 0


  def __init__(self, nome, preco, id = contador):
    Produto.contador += 1
    self.__id = Produto.contador
    self.__nome = nome
    self.__preco = preco

    #Getters
  def get_id(self):
    return self.__id

  def get_nome(self):
    return self.__nome

  def get_preco(self):
    return self.__preco

    #Setters
  def set_preco(self, novo_preco):
    if novo_preco <= 0:
      print(f"Valor inválido!")
    else:
      self.__preco = novo_preco

    #Função para desconto
  def aplicar_desconto(self, taxa):
    if taxa <= 0 and taxa >= 100:
      print(f"Desconto inválido!")
    else:
      desconto = self.__preco * taxa
      self.__preco -= desconto

    #Str para descrever
  def descricao(self):
    print(f"{self.__nome}, seu ID é {self.__id} e o seu preço é {self.__preco}")
```

* Crie ProdutoFisico herdando de Produto, com peso, e ProdutoDigital herdando de Produto, com tamanho_mb.
 * Em Produto, crie calcular_frete() retornando 0. Redefina-o em ProdutoFisico para retornar R$ 5,00 + R$ 2,00 por kg. ProdutoDigital deve manter frete igual a 0. Essa chamada comum deve demonstrar polimorfismo.

 * Depois de fechado, o pedido não deve aceitar novos produtos. Essa regra deve ser controlada internamente pela classe.
 * calcular_total() deve somar os preços dos produtos. calcular_frete_total() deve chamar calcular_frete() de cada item sem testar se ele é físico ou digital.
 * No programa principal, crie um cliente, ao menos três produtos (misturando físicos e digitais), monte um pedido, exiba o resumo, feche-o e tente adicionar mais um produto para comprovar a regra.

```python
# Definindo classe
class Produto:
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    # Getters
    def get_id(self):
        return self.__id

    def get_nome(self):
        return self.__nome

    def get_preco(self):
        return self.__preco

    # Setters
    def set_preco(self, novo_preco):
        if novo_preco <= 0:
            print("Valor inválido!")
        else:
            self.__preco = novo_preco

    # Calcular Frete
    def calcular_frete(self):
        return 0

    def get_frete(self):
        return 0

    # Função para desconto
    def aplicar_desconto(self, taxa):
        if taxa <= 0 or taxa >= 100:
            print("Desconto inválido!")
        else:
            desconto = self.__preco * (taxa / 100)
            self.__preco -= desconto

    # Descrição
    def descricao(self):
        print(f"{self.__nome}, seu ID é {self.__id} e o seu preço é {self.__preco}")
```

```python
# Herança - Produto Físico
class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.__peso = peso
        self.__frete = 0
        self.calcular_frete()

    # Getter
    def get_peso(self):
        return self.__peso

    # Calculo de frete
    def calcular_frete(self):
        self.__frete = 5.00 + (2.00 * self.__peso)

    def get_frete(self):
        return self.__frete
```

```python
# Herança - Produto Digital
class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.__tamanho_mb = tamanho_mb
        self.__frete = 0

    def get_tamanho_mb(self):
        return self.__tamanho_mb

    # Frete
    def get_frete(self):
        return self.__frete
```

```python
# Definindo classe
class Cliente:
    def __init__(self, nome, gmail):
        self.__nome = nome
        self.__gmail = gmail

    # Getters
    def get_nome(self):
        return self.__nome

    def get_gmail(self):
        return self.__gmail

    # STR de descrição
    def cliente_descricao(self):
        print(f"{self.get_nome()}, seu email é {self.get_gmail()}")
```

```python
# Definindo a classe
class Pedido:
    numerador = 0

    def __init__(self, cliente):
        Pedido.numerador += 1
        self.__numero = Pedido.numerador
        self.__cliente = cliente
        self.__produtos = []
        self.__fechado = False

    # Adicionar produto à lista
    def adicionar_produto(self, produto):
        if not self.__fechado:
            self.__produtos.append(produto)
        else:
            print("Pedido já foi fechado.")

    # Calcular o preço total dos produtos
    def calcular_total(self):
        total = 0
        for produto in self.__produtos:
            total += produto.get_preco()
        return total

    # Calcular o frete total
    def calcular_frete_total(self):
        frete_total = 0
        for produto in self.__produtos:
            frete_total += produto.get_frete()
        return frete_total

    # Fechamento de pedido
    def fechar_pedido(self):
        self.__fechado = True

    # SRT para visualização final
    def get_pedido(self):
        print(f"\n--- Pedido n° {self.__numero} ---")
        print(f"Cliente: {self.__cliente.get_nome()} ({self.__cliente.get_gmail()})")
        print("\nProdutos no Carrinho:")
        for produto in self.__produtos:
            print(f"  • Nome: {produto.get_nome()} | ID: {produto.get_id()} | Preço: R$ {produto.get_preco():.2f} | Frete: R$ {produto.get_frete():.2f}")

        print(f"\nSubtotal Produtos: R$ {self.calcular_total():.2f}")
        print(f"Frete Total: R$ {self.calcular_frete_total():.2f}")
        print(f"Valor Total Final: R$ {(self.calcular_total() + self.calcular_frete_total()):.2f}")
        print(f"Status: {'Fechado' if self.__fechado else 'Aberto'}")
        print("-" * 35)
```
